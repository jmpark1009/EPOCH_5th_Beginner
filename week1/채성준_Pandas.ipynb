```python
# 시리즈(Series)
# 1차원 배열의 값에 각 값에 대응되는 인덱스 부여 가능

sr = pd.Series([17000, 18000, 1000, 5000],
               index=["피자", "치킨", "콜라", "맥주"])
print('시리즈 출력 :')
print('-'*15)
print(sr)
```

```python
# 값과 인덱스를 출력할 수도 있음
print('시리즈의 값 : {}'.format(sr.values))
print('시리즈의 인덱스 : {}'.format(sr.index))
```

```python
# 데이터프레임(DataFrame)
# 2차원 리스트를 매개변수로 전달하는 기능
# 행방향 인덱스(index)와 열방향 인덱스(column)가 존재

values = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
index = ['one', 'two', 'three']
columns = ['A', 'B', 'C']

df = pd.DataFrame(values, index=index, columns=columns)

print('데이터프레임 출력 :')
print('-'*18)
print(df)
```

```python
# 인덱스, 값, 열을 각각 출력해보기

print('데이터프레임의 인덱스 : {}'.format(df.index))
print('데이터프레임의 열이름: {}'.format(df.columns))
print('데이터프레임의 값 :')
print('-'*18)
print(df.values)
```

```python
# 데이터 프레임 생성하기
# 딕셔너리를 사용해서 데이터 프레임 만들기
# 원래는 리스트, 시리즈, 딕셔너리, ndarrays 등도 가능

# 리스트로 생성하기
data = [
    ['1000', 'Steve', 90.72],
    ['1001', 'James', 78.09],
    ['1002', 'Doyeon', 98.43],
    ['1003', 'Jane', 64.19],
    ['1004', 'Pilwoong', 81.30],
    ['1005', 'Tony', 99.14],
]

df = pd.DataFrame(data)
print(df)
```

```python
# 딕셔너리를 통해 테이터 프레임 만들어보기

# 딕셔너리로 생성하기
data = {
    '학번' : ['1000', '1001', '1002', '1003', '1004', '1005'],
    '이름' : [ 'Steve', 'James', 'Doyeon', 'Jane', 'Pilwoong', 'Tony'],
    '점수': [90.72, 78.09, 98.43, 64.19, 81.30, 99.14]
    }

df = pd.DataFrame(data)
print(df)
```

```python
# Numpy 슬라이싱: 파이썬의 자료구조인 리스트처럼 슬라이싱 기능 지원

mat = np.array([[1, 2, 3], [4, 5, 6]])
print(mat)
```

```python
# 첫번째 행 출력
slicing_mat = mat[0, :]
print(slicing_mat)
```

```python
# 두번째 열 출력
slicing_mat = mat[:, 1]
print(slicing_mat)
```

```python
# Numpy 정수 인덱싱(integer indexing)
# 연속적이지 않은 원소로 배열을 만들 경우, 슬라이싱으로는 만들 수 없음

# 인덱싱을 사용하여 배열을 구성할 수 있음
# 인덱싱: 원하는 위치의 원소들을 뽑는 기능

mat = np.array([[1, 2], [4, 5], [7, 8]])
print(mat)
```

```python
# 1행 0열의 원소
# => 0부터 카운트하므로 두번째 행 첫번째 열의 원소

print(mat[1, 0])
```

```python
# mat[[2행, 1행],[0열, 1열]]
# 각 행과 열의 쌍을 매칭하면 2행 0열, 1행 1열의 두 개의 원소

indexing_mat = mat[[2, 1],[0, 1]]
print(indexing_mat)
```

```python
# Numpy 연산: Numpy를 사용하면 배열간 연산을 손쉽게 수행할 수 있
# 연산자 +, -, *, /를 사용하는 방법이 있고
# np.add(), np.subtract(), np.multiply(), np.divide()를 사용할 수도 있다

x = np.array([1,2,3])
y = np.array([4,5,6])
```

```python
# result = np.add(x, y)와 동일

result = x + y
print(result)
```

```python
# result = np.subtract(x, y)와 동일

result = x - y
print(result)
```

```python
# result = np.multiply(result, x)와 동일

result = result * x
print(result)
```

```python
# result = np.divide(result, x)와 동일

result = result / x
print(result)
```

```python
# 위에서 *를 통해 수행한 것은 요소별 곱
# 벡터와 행렬곱 또는 행렬곱을 위해서는 dot()을 사용

mat1 = np.array([[1,2],[3,4]])
mat2 = np.array([[5,6],[7,8]])
mat3 = np.dot(mat1, mat2)
print(mat3)
```

# 파이썬을 이용한 날짜와 시간 다루기

**datetime** = 시계열 데이터를 다루는 다양한 상황에서 사용

그래서 스트링(String) 타입의 문자를 날짜 형태 포맷(format)으로 바꾸거나  시간 단위 변경, 시각화 등에 응용

**time** : 시간 기능만 제공 (시, 분, 초, 마이크로초)

**date** : 날짜 기능만 제공 (연, 월, 일)

**datetime** : 날짜와 시간 기능을 제공 (- date + time)

**timedelta** : datetime 인스턴스 간의 차이 구



```python
import datetime

# strptime() 함수에 datetime_str 변수와 형식을 매개변수로 전달
datetime_str = '2018-05-13 12:34:56'
```

```python
format = '%Y-%m-%d %H:%M:%S'

#  str 타입의 문자가 datetime 타입으로 변환됨
datetime_dt = datetime.datetime.strptime(datetime_str, format)
print(datetime_dt)
```

```python
# datetime을 str타입의 문자로 변환

datetime_str = datetime_dt.strftime('%Y-%m-%d %H:%M:%S')

print(type(datetime_str)) # <class 'str'>
print(datetime_str)       # 2018-08-24 18:00:25
```

```python
# 분, 초를 제외하고 시간까지만 출력

datetime_str = datetime_dt.strftime('%Y-%m-%d %H')

print(type(datetime_str)) # <class 'str'>
print(datetime_str)       # 2018-08-24 18:00:25
```

```python
# 년도와 달까지만 출력

datetime_str = datetime_dt.strftime('%Y-%m')

print(type(datetime_str)) # <class 'str'>
print(datetime_str)       # 2018-08-24 18:00:25
```

```python
# String을 datetime으로 변환할 때는 strptime() 함수를 사용
# 그 반대의 경우, strftime() 함수를 사용
```
