2.6 조건 연산

```python
arr = np.array([10, 20, 30])
print(arr > 10)
```

```python
arr = np.array([10, 20, 30])
cond = [False, True, True]
print(arr[cond])
```

```python
arr = np.array([10, 20, 30])
print(arr[[False, True, True]])
```

```python
arr = np.array([10, 20, 30])
cond = arr > 10
print(arr[cond])
```

```python
print(arr[ arr > 10 ])
```

```python
cond = np.array([False, True, True])
print(cond.all())
print(cond.any())
```

```python
arr = np.array([10, 20, 30])
cond0 = arr > 10
cond1 = arr < 30
print(arr[ (arr > 10) | (arr < 30) ])
```

```python
arr = np.array([10, 20, 30])
cond = arr > 10
arr[cond] = 1
arr[~cond] = 0
print(arr)
```

```python
arr = np.array([10, 20, 30,])
arr = np.where(arr > 10, 1, 0)
print(arr)
```

```python
arr = np.array([10, 20, 30])
cond = arr > 10
arr[cond] = arr[cond] + 10
arr[~cond] = arr[~cond] - 10
print(arr)
```

```python
arr = np.array([10, 20, 30])
arr = np.where(arr > 10, arr + 10, arr - 10)
print(arr)
```

```python
lge = np.array([93000, 82400, 99100, 81000, 72300])
days = lge[lge <= 85000]
print(len(days))
```
