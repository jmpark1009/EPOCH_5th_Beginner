```python
# 1. 판다스(Pandas)
import pandas as pd
```

```python
# 1) 시리즈(Series)
# 시리즈 클래스: 1차원 배열의 값(values)에 각 값에 대응되는 인덱스(index)를 부여할 수 있는 구조
sr = pd.Series([17000, 18000, 1000, 5000],
               index = ["피자", "치킨", "콜라", "맥주"])
print('시리즈 출력 : ')
print('-' * 15)
print(sr)
```

```python
# 값, 인덱스를 출력
print('시리즈의 값 : {}'.format(sr.values))
print('시리즈의 인덱스 : {}'.format(sr.index))
```

```python
# 2) 데이터프레임(DataFrame)
# 데이터프레임: 2차원 리스트를 매개변수로 전달
# 2차원이므로 행방향 인덱스와 열방향 인덱스가 존재
# 행과 열을 가지는 자료구조임
# 시리즈 -> 인덱스와 값으로 구성
# 데이터프레임 -> 열, 인덱스, 값으로 구성
```

```python
values = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
index = ['one', 'two', 'three']
columns = ['A', 'B', 'C']

df = pd.DataFrame(values, index=index, columns=columns)

print('데이터프레임 출력 :')
print('-'*18)
print(df)
```

```python
# 생성된 데이터프레임으로부터 인덱스, 값, 열을 각각 출력
print('데이터프레임의 인덱스 : {}'.format(df.index))
print('데이터프레임의 열이름: {}'.format(df.columns))
print('데이터프레임의 값 :')
print('-'*18)
print(df.values)
```

```python
# 3) 데이터프레임의 생성
# 데이터프레임은 리스트, 시리즈, 딕셔너리, Numpy의 ndarrays, 또 다른 데이터프레임으로부터 생성 가능
```

```python
# 생성된 데이터프레임에 열을 지정할 수 있음
# 열이름을 지정하고 출력
df = pd.DataFrame(data, columns=['학번', '이름', '점수'])
print(df)
```

```python
# 1) np.array()
# numpy의 핵심 -> ndarray
# np.array()는 리스트, 튜플, 배열로 부터 ndarray를 생성
```

```python
# 리스트를 가지고 1차원 배열을 생성하기
vec = np.array([1, 2, 3, 4, 5])
print(vec)
```

```python
# 2차원 배열 만들기
mat = np.array([[10, 20, 30], [ 60, 70, 80]])
print(mat)
# array() 안에 하나의 리스트만 들어가므로
# 리스트의 리스트를 넣어야 함
```

```python
# 두 배열의 타입을 확인하기
print('vec의 타입 :',type(vec))
print('mat의 타입 :',type(mat))
# 동일하게 타입이 numpy.ndarray라고 나오게 됨
# numpy 배열에는축의 개수와 크기라는 개념이 존재
# 배열의 크기를 정확히 숙지하는 것은 딥러닝에서 매우 중요함
```

```python
print('vec의 축의 개수 :',vec.ndim)
print('vec의 크기(shape) :',vec.shape)
```

```python
print('mat의 축의 개수 :',mat.ndim)
print('mat의 크기(shape) :',mat.shape)
```

```python
# 2) ndarray의 초기화
# ndarray를 만드는 다양한 다른 방법이 존재
```

```python
# np.zeros()는 배열의 모든 원소에 0을 삽입
# 모든 값이 0인 2X3 배열 생성
zero_mat = np.zeros((2,3))
print(zero_mat)
```

```python
# np.ones()는 배열의 모든 원소에 1을 삽입
# 모든 값이 1인 2x3 배열 생성
one_mat = np.ones((2,3))
print(one_mat)
```

```python
# np.full()은 배열에 사용자가 지정한 값을 삽입
# 모든 값이 특정 상수인 배열 생성(이 경우 7)
same_value_mat = np.full((2,2), 7)
print(same_value_mat)
```

```python
# np.eye()는 대각선으로는 1이고 나머지는 0인 2차원 배열을 생성
# 대각선 값이 1이고 나머지 값이 0인 2차원 배열을 생성
eye_mat = np.eye(3)
print(eye_mat)
```

```python
# np.random.random()은 임의의 값을 가지는 배열을 생성
# 임의의 값으로 채워진 배열 생성
random_mat = np.random.random((2,2)) # 임의의 값으로 채워진 배열 생성
print(random_mat)
```

```python
# 3) np.arange()
```

```python
# np.arange(n)은 0부터 n-1까지의 값을 가지는 배열을 생성
# 0부터 9까지
range_vec = np.arange(10)
print(range_vec)
```

```python
# np.arange(i, j, k)는 i부터 j-1까지 k씩 증가하는 배열을 생성
# 1부터 9까지 +2씩 적용되는 범위
n = 2
range_n_step_vec = np.arange(1, 10, n)
print(range_n_step_vec)
```

```python
# 4) np.reshape()
```

```python
# np.reshape()은 내부 데이터는 변경하지 않으면서
# 배열의 구조를 바꿈
# 0부터 29까지의 숫자를 생성하는 arange(30)을 수행한 후
# 원소의 개수가 30개이므로 5행 6열의 행렬로 변경하기
reshape_mat = np.array(np.arange(30)).reshape((5,6))
print(reshape_mat)
```

```python
# 5) Numpy 슬라이싱
```

```python
# ndarray를 통해 만든 다차원 배열은 파이썬의 자료구조인 리스트처럼 슬라이싱 기능을 지원함
# 슬라이싱 기능을 사용하여 특정 행이나 열들의 원소들을 접근할 수 있음
mat = np.array([[1, 2, 3], [4, 5, 6]])
print(mat)
```

```python
# 첫번째 행 출력
slicing_mat = mat[0, :]
print(slicing_mat)
```

```python
# 두번째 열 출력
slicing_mat = mat[:, 1]
print(slicing_mat)
```

```python
# 6) Numpy 정수 인덱싱(integer indexing)
```

```python
# 슬라이싱을 사용하면 배열로부터 부분 배열을 추출할 수 있지만
# 연속적이지 않은 원소로 배열을 만들 경우에는 슬라이싱으로 만들 수 없음
# 예를 들어 2행 2열의 원소와
# 5행 5열의 원소를 뽑아서 하나의 배열로 만들고자 하는 경우
# 이런 경우에는 인덱싱을 사용하여 배열을 구성할 수 있음
# 인덱싱은 원하는 위치의 원소들을 뽑을 수 있음
mat = np.array([[1, 2], [4, 5], [7, 8]])
print(mat)
```

```python
# 특정 위치의 원소만을 가져와보기
# 1행 0열의 원소
# => 0부터 카운트하므로 두번째 행 첫번째 열의 원소.
print(mat[1, 0])
```

```python
# 특정 위치의 원소 두 개를 가져와 새로운 배열을 만들어보기
# mat[[2행, 1행],[0열, 1열]]
# 각 행과 열의 쌍을 매칭하면 2행 0열, 1행 1열의 두 개의 원소
indexing_mat = mat[[2, 1],[0, 1]]
print(indexing_mat)
```

```python
# 7) Numpy 연산
```

```python
# Numpy를 사용하면 배열간 연산을 손쉽게 수행할 수 있음
# 덧셈, 뺄셈, 곱셈, 나눗셈을 위해서는 연산자
# +, -, *, /를 사용할 수 있음
# 또는 np.add(), np.substract(), np.multiply(), np.divide()를 사용할 수 있음
```

```python
x = np.array([1,2,3])
y = np.array([4,5,6])
```

```python
# result = np.add(x, y)와 동일
result = x + y
print(result)
```

```python
# result = np.subtract(x, y)와 동일
result = x - y
print(result)
```

```python
# result = np.multiply(result, x)와 동일
result = result * x
print(result)
```

```python
# result = np.divide(result, x)와 동일
result = result / x
print(result)
```

```python
# *를 통해 수행한 것은 요소별 곱
# Numpy에서 벡터와 행렬곱 또는 행렬곱을 위해서는 dot()을 사용해야 함
mat1 = np.array([[1,2],[3,4]])
mat2 = np.array([[5,6],[7,8]])
mat3 = np.dot(mat1, mat2)
print(mat3)
```

## 01) datetime
- time: 시간 기능만 제공(시, 분, 초, 마이크로초)
  - 예) datetime.time(hour=4, minute=3, second=10, microsecond=1000)
- date: 날짜 기능만 제공(연, 월, 일)
  - 예) datetime.date(year=2019, month=1, day=10)
- datetime: 날짜와 시간 기능을 제공(date + time)
  - 예) datetime.datetime(year=2019, month=10, day=24, hour=4, minute=3, second=10, microsecond=1000)
- timedelta: datetime 인스턴스 간의 차이를 구함
  - 예) datetime.datetime(year=2019, month=10, day=24, hour=4, minute=3, second=10, microsecond=1000)

- 실제 분석에서 날짜와 시간은 문자열 타입으로 주어지는 경우가 많음
- 따라서 자유롭게 데이터 타입을 변환하는 방법을 익혀야함

```python
# 스트링(String) 타입의 문자를 날짜 형태 포맷으로 바꾸거나
# 시간 단위 변경, 시각화 등에 응용할 수 있음
# datetime에는 time, date, datetime, timedelta 외에도 여러 기능이 있음
```

```python
# str 타입의 문자를 datetime 형식으로 바꾸는 방법
import datetime
datetime_str = '2018-05-13 12:34:56'
```

```python
# strptime()함수에 datetime_str변수와 형식을 매개변수로 전달
# 입력한 str 타입의 날짜와 날짜를 표현하는 형식을 strtime()함수에 전달
# => str 타입의 문자가 datetime 타입으로 변환한 것을 알 수 있음
format = '%Y-%m-%d %H:%M:%S'
datetime_dt = datetime.datetime.strptime(datetime_str, format)
print(datetime_dt)

print(type(datetime_dt))
```

```python
# datetime을 str타입의 문자로 변환
# datetime_dt -> 현재 datetime 타입
datetime_str = datetime_dt.strftime('%Y-%m-%d %H:%M:%S')

print(type(datetime_str))
print(datetime_str)
```

```python
# strptime()함수를 호출할 때 변환된 datetime_dt에서 strftime() 함수를 호출
# 함수 매개변수에는 변환하려는 날짜, 시간 형식을 전달함
# str 타입으로 변환된 것과 우리가 전달한 형식에 맞게 값이 나온 것을 확인할 수 있음
# 또 다른 형식으로 변경하려면 다음과 같이 수정해서 전달하면 됨

# 분, 초를 제외하고 시간까지만 출력하기
datetime_str = datetime_dt.strftime('%Y-%m-%d %H')

print(type(datetime_str))
print(datetime_str)
```

```python
# 년도와 달까지만 출력하기
datetime_str = datetime_dt.strftime('%Y-%m')

print(type(datetime_str))
print(datetime_str)
```

```python
# 정리
# '%Y-%m-%d %H' -> 연-월-일-시 형식으로 출력
#  '%Y-%m' -> 연-월 출력

# strptime()함수와 strftime()함수는 서로 반대되는 기능을 가지고 있음
# strptime() -> str 타입으로 된 날짜 데이터를 datetime 형태로 변환
# strftime() -> datetime 형태의 데이터를 str 타입으로 변환

# 주의해야 할 점: 형식을 정의할 때 지정한 형식이 대소문자를 구분하므로 형식을 맞추거나 정확한 날짜를 표현해야 함
# 참고자료: https://dojang.io/mod/page/view.php?id=2463
```

## 02) 날짜와 시간을 다루는 넘파이(Numpy)
- 참고자료: 클래스와 객체( https://wikidocs.net/28 )
- 넘파이에서는 날짜와 시간을 datetime64 객체로 표현
- 넘파이의 datetime64와 파이썬의 datetime.datetime과 차이점
  - datetime -> 시간을 마이크로초 단위까지 관리
  - 넘파이의 datetime64는 아토초 단위까지 관리

- 넘파이의 날짜와 시간을 이해해야 하는 중요한 이유
  - 우리가 궁극적으로 사용할 판다스의 날짜와 시간 함수가 넘파이를 기반으로 만들어졌기 때문

```python
# 1. datetime64
# Numpy 객체를 이용해 날짜를 생성하는 방식에는 두 가지 방식이 있음
# (1) 문자열 형식으로 전달해서 생성
import numpy as np
np.datetime64('2019-01-01')

# (2) 유닉스 시간을 이용한 방식
# 첫번째 입력 매개변수 -> 수치데이터
# 두번째 입력 매개변수 -> 단위
# ns -> 나노초 단위
# D -> 일 단위
# s -> 초 단위
import numpy as np

# 나노초 단위
np.datetime64(1000,'ns')
```

```python
# 수치 바꿔보기
np.datetime64(10000000000,'ns')
```

```python
# 초 단위로 출력하기
np.datetime64(1000000000,'s')
```

```python
# 일 단위로 출력하기
np.datetime64(10000,'D')
```

```python
# 2. array()
# np.array() 함수에 str타입의 날짜 리스트를 전달하고
# dtype(데이터 타입)을 'datetime64'로 설정
```

```python
np.array(['2007-07-13', '2006-01-13', '2010-08-13'], dtype='datetime64')
```

```python
# dtype에서 datetime64를 적을 때
# []안에 어떤 값을 넣느냐에 따라서 날짜 단위를 지정할 수도 있음

# 달단위
np.array(['2007-07-13', '2006-01-13', '2010-08-13'], dtype='datetime64[M]')
```

```python
# 3. arange()
```

```python
# 달단위
np.arange('2005-02', '2006-03', dtype='datetime64[M]')
```

```python
# 주단위
np.arange('2005-02', '2006-03', dtype='datetime64[W]')
```

```python
# 일단위
np.arange('2005-02', '2005-03', dtype='datetime64[D]')
```

```python
# 시간단위
np.arange('2005-02-01', '2005-02-02', dtype='datetime64[h]')
```

```python
# 4. datetime64를 이용한 날짜 차이 구하기
# 각각 날짜를 생성할 때 어떤 단위를 전달하는지에 따라서 생성되는 범위가 달라짐
# 두 날짜의 간격을 구할 때는 단순하게 빼주면 됨
# 직관적인 날짜 간격을 구하기 쉬움
```

```python
np.datetime64('2009-01-01') - np.datetime64('2008-01')
# 366이라는 숫자가 나옴
```

```python
np.datetime64('2009-01-01') - np.datetime64('2008-01-01')
# 동일한 결과가 나옴
# 2008년 1월을 했을 때 자동으로 2008년 1월 1일로 인식하여 366일의 차이가 나온 것
```

```python
np.datetime64('2009') - np.datetime64('2008-01')
# 12라는 숫자가 나옴
# 2009년 1월과 2008년 1월의 차이가 12개월이라는 의미
```

```python
# 날짜 간격을 구할 때 두 날짜 중 작은 단위 날짜로 맞춤
# Y : 연
# M : 월
# W : 주
# D : 일
# h : 시간
# m : 분
# s : 초
# ms : 밀리초
# us : 마이크로초
# ns : 나노초
# ps : 피코초
# fs : 펨토초
```
