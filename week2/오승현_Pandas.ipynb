```python
from sklearn import datasets
iris_data=datasets.load_iris()
```

```python
df_data=pd.DataFrame(iris_data['data'],columns=iris_data['feature_names'])
df_data.sample(5)
```

```python
df_target=pd.DataFrame(iris_data['target'],columns=['species'])
df_target.sample(5)
```

```python
df_target['species'].unique()
```

```python
df=pd.concat([df_data,df_target],axis=1)
df
```

```python
cols=list(df.columns)
corr=df[cols].corr(method='pearson')
corr
```

```python
a = sns.heatmap(corr.values,
                cbar=True,
                annot=True,
                fmt='.2f',
                annot_kws={'size':15},
            yticklabels=cols,
            xticklabels=cols)
plt.show()
```

```python
sns.pairplot(df[cols])
plt.show()
```

```python
q=df['species'].value_counts().index.tolist()
w=df['species'].value_counts().values.tolist()
plt.pie(w,labels=q,autopct='%.02f%%')
plt.show()
```
