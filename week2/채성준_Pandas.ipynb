# 4-5) AND, OR, NOT 연산자

```python
import pandas as pd

url = 'https://raw.githubusercontent.com/justmarkham/DAT8/master/data/drinks.csv'
drink_df = pd.read_csv(url)
drink_df
```

```python
# AND, OR, NOT
# & | ~

# NOT 조건
# continent의 값이 'EU'가 아닌 경우만을 필터링하는 코드
drink_df[~(drink_df.continent=='EU')]
```

```python
# AND 조건
# ontinent의 값이 'EU'이면서 wine_servings의 값이 300보다 큰 경우
drink_df[(drink_df.continent=='EU') & (drink_df.wine_servings > 300)]
```

```python
# len(입력)을 사용하면 입력의 길이를 계산하여 알려줌(행길이 출력)

# OR 조건
len(drink_df[(drink_df.continent=='EU') | (drink_df.wine_servings > 300)])
```

## 4-7) 상관 관계 분석

**상관 분석**이란 두 변수 간의 선형적 관계를 상관 계수로 표현하는 것

```python
# Matplotlib = 파이썬에서 자료를 차트나 플롯으로 시각화하는 패키지
# Seaborn = Matplotlib을 기반으로 다양한 테마와 기능을 추가한 시각화 패키지

import pandas as pd # 판다스 임포트.
import matplotlib.pyplot as plt # 맷플롯립 임포트.
import seaborn as sns # 시본 임포트
```

```python
# 다수의 파이썬 리스트를 통해 데이터프레임을 만드는 방법
# pd.DataFrame({"열이름1":리스트1, "열이름2":리스트2, "열이름3":리스트3})

test_df = pd.DataFrame({"v1":[100,200,300,400], "v2":[400,200,100,250], "v3":[40,60,60,100]})
test_df
```

```python
# 데이터의 상관계수를 구하는 방법
# 데이터프레임의 이름.corr(method='pearson')

corr = test_df.corr(method='pearson')
corr
```

```python
# 히트맵 그리는 방법

corr.values
```

```python
column_names = ['ver1', 'ver2', 'ver3']
```

```python
# 상관계수 차트 그릴 때는 seaborn의 heatmap()을 주로 사용
# sns.heatmap(데이터프레임의 상관계수 데이터)

# 레이블의 폰트 사이즈를 조정
sns.set(font_scale=2.0)

test_heatmap = sns.heatmap(corr.values, # 데이터
                          cbar = False, # 오른쪽 컬러 막대 출력 여부
                           annot = True, # 차트에 숫자를 보여줄 것인지 여부
                          annot_kws={'size' : 20}, # 숫자 출력 시 숫자 크기 조절
                           fmt = '.2f', # 숫자의 출력 소수점자리 개수 조절
                           square = 'True', # 차트를 정사각형으로 할 것인지
                          yticklabels=column_names, # y축에 컬럼명 출력
                          xticklabels=column_names) # x축에 컬럼명 출력
plt.tight_layout()
plt.show()
```

```python
# 레이블의 폰트 사이즈를 조정
sns.set(font_scale=2.0)

test_heatmap = sns.heatmap(corr.values,
                           cbar = True,
                           annot = True,
                           annot_kws={'size' : 20},
                           fmt = '.2f',
                           square = True,
                          yticklabels=column_names,
                          xticklabels=column_names)
plt.tight_layout()
plt.show()
```

```python
# 예제 : fmt = '.2f'의 값을 fmt = '.3f'으로 바꿔서 재실행

# 레이블 폰트 사이즈 조정
sns.set(font_scale=2.0)

test_heatmap = sns.heatmap(corr.values,
                           cbar = True,
                           annot = True,
                           annot_kws={'size' : 20},
                           fmt = '.3f', # 셋째 소수점 자리까지 나오게 된다
                           square = True,
                          yticklabels=column_names,
                          xticklabels=column_names)
plt.tight_layout()
plt.show()
```

```python
# 산점도(scatter plot) = 좌표사에 점들을 표시하는 방법
# 두 개 벼수 간의 관계를 나타내는 데에 용이하다
# 방식: sns.pairplot(데이터프레임)

# whitegrid = 배경에 하얀 선을 넣는다.
sns.set(style='whitegrid')
sns.pairplot(test_df)
plt.show()
```

## 4-8) 응용! 상관 관계 분석과 산점도



```python
import pandas as pd

url = 'https://raw.githubusercontent.com/justmarkham/DAT8/master/data/drinks.csv'
drink_df = pd.read_csv(url)
drink_df
```

```python
# 'beer_servings', 'wine_servings' 두 피처간의 상관계수 계산
# pearson은 상관계수를 구하는 계산 방법 중 하나를 의미
# 가장 널리 쓰이는 방법입니다.

corr = drink_df[['beer_servings', 'wine_servings']].corr(method = 'pearson')
corr
```

```python
hm = sns.heatmap(corr.values,
            cbar=True,
            annot=True,
            square=True,
            fmt='.2f',
            annot_kws={'size': 15},
            yticklabels=column_names,
            xticklabels=column_names,
            cmap='Blues') # <- 이 부분을 추가해 보세요!
```

## 4-9) 데이터 탐색하기(결측값 제거, 시각화, 통계 확인)

```python
import pandas as pd

url = 'https://raw.githubusercontent.com/justmarkham/DAT8/master/data/drinks.csv'
drink_df = pd.read_csv(url)
drink_df
```

```python
# 결측값 제거 방법
# 1. 아예 제거시킨다 = dropna()
# 2. 특정값으로 채워주거나 = fillna()

# 이 경우 기타 값이라는 의미에서 'ETC'를 넣어준다.
drink_df['continent'] = drink_df['continent'].fillna('ETC')
drink_df.sample(10)
```

```python
# 결측값 사라졌는지 확인하기 = insull().sum()
drink_df.isnull().sum()
```

```python
# 파이차트 그리기 조건
# 1. 파이 차트로 사용할 데이터의 이름이 담긴 리스트
# 2. 해당 이름에 해당하는 데이터의 값이 담긴 리스트

# continent 열에 대한 결과 출력 value_counts()
drink_df['continent'].value_counts()
```

```python
# continent 열의 타입 확인
type(drink_df['continent'].value_counts())
# 시리즈로 나왔
```

```python
# 인덱스 = continent 이름
# 실제 값 부분 = 해당 continent에 있는 값들을 각각 카운트한 값
# values로 접근하므로 각각 접근 후에 리스트로 값을 변환하는 tolist()를 사용

pie_labels = drink_df['continent'].value_counts().index.tolist()
pie_values = drink_df['continent'].value_counts().values.tolist()
print(pie_labels)
print(pie_values)
```

```python
# 파이차트를 그리는 방법 = plt.pie(데이터의 실질적인 값, labels=데이터의 레이블 리스트)

plt.pie(pie_values, labels=pie_labels, autopct='%.02f%%')
# autopct는 ????
plt.title('Percentage of each continent')
plt.show()
```

```python
# Group By = 특정값 기준 그룹핑한 후
# 그룹 별로 통계적 수치 정보를 구할 수 있는 방법
# 데이터프레임의 이름.groupby('그룹핑 기준이 되는 열')['보고자 하는 열'].통계 함수

# 어느 대륙이 평균적으로 맥주를 더 먹을까?
drink_df.groupby('continent')['beer_servings'].mean()
```

```python
# 응용
# 전체 평균보다 많은 알코올을 섭취하는 대륙 출력

total_mean = drink_df.total_litres_of_pure_alcohol.mean()
continent_mean = drink_df.groupby('continent')['total_litres_of_pure_alcohol'].mean()
continent_over_mean = continent_mean[continent_mean >= total_mean]
print(continent_over_mean)
```

```python
# wine_servings 값이 가장 높은 대륙은? #idxmax()

beer_continent = drink_df.groupby('continent').wine_servings.mean().idxmax()
print(beer_continent)
```
