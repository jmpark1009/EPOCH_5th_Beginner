```python
y=np.arange(3)
years=['2018','2019','2020']
values=[100,400,900]
plt.barh(y, values)
plt.yticks(y, years)

plt.show()
```

```python
y=np.arange(3)
years=['2018','2019','2020']
values=[100,400,900]
i=0
d=['y','dodgerblue','C2','#e35f62']
fig, ax=plt.subplots(2,2)
for a in [0,1]:
    for b in [0,1]:
        ax[a,b].barh(y, values, color=d[i])
        ax[a,b].set_yticks(y)
        ax[a,b].set_yticklabels(years)
        i+=1
plt.show()
```

```python
colors = ['y', 'dodgerblue', 'C2']
plt.barh(y, values, color=colors)
plt.yticks(y, years)

plt.show()
```

```python
y=np.arange(3)
years=['2018','2019','2020']
values=[100,400,900]
i=0
d=np.arange(0.4,1.2,0.2)
fig, ax=plt.subplots(2,2)
for a in [0,1]:
    for b in [0,1]:
        ax[a,b].barh(y, values, height=d[i])
        ax[a,b].set_yticks(y)
        ax[a,b].set_yticklabels(years)
        i+=1
plt.show()
```

```python
plt.barh(y, values, align='edge', edgecolor='#eee', linewidth=5, tick_label=years)
plt.show()
```
